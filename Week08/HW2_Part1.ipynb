In class we solved a **regression** task with the mean squared error. Here the task is **binary classification**, and the loss is the binary cross-entropy. PyTorch has two versions of it; read their documentation:

* [`nn.BCELoss`](https://pytorch.org/docs/stable/generated/torch.nn.BCELoss.html) takes **probabilities**, so the model must end with a sigmoid;
* [`nn.BCEWithLogitsLoss`](https://pytorch.org/docs/stable/generated/torch.nn.BCEWithLogitsLoss.html) takes **logits** (unnormalized scores) and applies the sigmoid inside, which is numerically more stable. With it, the model has no final sigmoid.

(For more than two classes, [`nn.CrossEntropyLoss`](https://pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html) plays the same role and also takes logits.)

## 0. The data (provided)

Use the `load_planar_dataset` function to create the data: 400 points in the plane, in two classes that form the petals of a "flower". No straight line separates them.

```python
def load_planar_dataset():
    np.random.seed(1)
    m = 400 # number of examples
    N = int(m/2) # number of points per class
    D = 2 # dimensionality
    X = np.zeros((m,D)) # data matrix where each row is a single example
    Y = np.zeros(m, dtype='uint8') # labels vector (0 for red, 1 for blue)
    a = 4 # maximum ray of the flower

    for j in range(2):
        ix = range(N*j,N*(j+1))
        t = np.linspace(j*3.12,(j+1)*3.12,N) + np.random.randn(N)*0.2 # theta
        r = a*np.sin(4*t) + np.random.randn(N)*0.2 # radius. 4 corresponds to the number of petals
        X[ix] = np.c_[r*np.sin(t), r*np.cos(t)]
        Y[ix] = j

    return X, Y
```
